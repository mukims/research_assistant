```python
from langchain_core.messages import AIMessage, HumanMessage, ToolMessage
from langgraph.graph import END, START, StateGraph
from langgraph.graph.message import add_messages 
from langchain_ollama import ChatOllama
from langchain_core.tools import tool 
from langgraph.prebuilt import ToolNode
```

```python
from typing import Annotated
from typing import TypedDict
from langgraph.graph.message import add_messages
```

```python
@tool
def get_language(city:str):
    """Get the language spoken in the city. """
    return f"The language spoken in the {city} is ::"

def get_language2(city: str) -> str:
    """Get the primary language spoken in a given city."""
    lookup = {
        "paris": "French",
        "tokyo": "Japanese",
        "berlin": "German",
        "london": "English",
        "madrid": "Spanish",
    }
    key = city.strip().lower()
    if key in lookup:
        return f"The primary language in {city} is {lookup[key]}."
    return f"I do not have record of the language spoken in '{city}'."

@tool
def compute_demographic_power(x: float, y: float) -> float:
    """Computes x**y, where:
    - x: Total number of cities in the country associated with the target city.
    - y: Portion of the country's population residing in that city (as a decimal fraction between 0.0 and 1.0).
    
    Call this tool ONLY when both x and y are known or can be reasonably estimated.
    """
    return round(float(x ** y), 4)
```

```python
tools = [get_language,compute_demographic_power]
```

```python
SYSTEM_PROMPT = (
    "You are a helpful AI assistant equipped with tools to analyze cities.\n\n"
    "When asked about a city:\n"
    "1. Language: Call `get_language(city)` to identify the primary language.\n"
    "2. Demographic Metric (x**y):\n"
    "   - x = total number of cities/communes in the country associated with the city.\n"
    "   - y = portion of the country's population living in that city (decimal between 0.0 and 1.0).\n"
    "3. Execution:\n"
    "   - If exact numbers are known, call `compute_demographic_power(x, y)` and explain the values.\n"
    "   - If exact numbers are unknown, make a logical estimate for x and y, state: "
    "'I do not have sufficient demographic data for [city], but my logical estimate is [reasoning]', "
    "and then call `compute_demographic_power(x, y)` with your estimated values.\n"
    "4. Final Output: Synthesize the language, demographic logic, and computed metric into a clear summary."
)
```

```python
llm = ChatOllama(model="qwen2.5:7b", temperature=0.1).bind_tools(tools)
```

```python
from langchain_core.messages import SystemMessage
class AgentState(TypedDict):
    messages: Annotated[list,add_messages]

def agent_node(state:AgentState):
    prompt_message = [SystemMessage(content=SYSTEM_PROMPT)] + state["messages"]
    response = llm.invoke(prompt_message)
    return {"messages":[response]}


tool_node = ToolNode(tools)

def should_continue(state:AgentState):
    last_message = state["messages"][-1]
    if last_message.tool_calls:
        return "tools"
    return END
```

```python
workflow = StateGraph(AgentState)



workflow.add_node("agent", agent_node)
workflow.add_node("tools", tool_node)


workflow.add_edge(START,"agent")

workflow.add_conditional_edges(
    "agent",
    should_continue,
    {
        "tools":"tools",
        END:END
    }
)


workflow.add_edge("tools","agent")



app = workflow.compile()
```
